## Handling Null Values

```python
## fill Null Values of country column
df["country"] = df["country"].fillna("Unknown")

## fill Null Values of cast column
df["cast"] = df["cast"].fillna("Not Listed")

## fill Null Values of director column
df["director"] = df["director"].fillna("Unknown")

## filling dates in blank column
import numpy as np

random_dates = np.random.choice(df["date_added"].dropna(), size=df["date_added"].isnull().sum())

df.loc[df["date_added"].isnull(), "date_added"] = random_dates

## fill Null Values of country column
df["rating"] = df["rating"].fillna("Unknown")

## fill Null Values of country column
df["duration"] = df["duration"].fillna("Not Defined")
```

```python
df.isnull().sum()
```

```python
## delete duplicate values

df.drop_duplicates(inplace = True)
```

```python
## change date in proper format

df["date_added"] = pd.to_datetime(
    df["date_added"],
    errors="coerce"
)
df["date_added"].head()
```

## Feature Engineering

### New features were created to improve analytical capability and trend analysis.

```python
df["content_age"] = 2026 - df["release_year"]
```

```python
df["decade"] = (df["release_year"] // 10) * 10
```

```python
df["month_added"] = df["date_added"].dt.month
```

```python
df["year_added"] = df["date_added"].dt.year
```

```python
## Creating new cleaned dataset csv file

df.to_csv("../data/cleaned/cleaned_netflix.csv", index=False)
print("CSV Saved!!!")
```
