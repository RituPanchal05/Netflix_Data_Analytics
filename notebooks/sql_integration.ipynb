# 3. SQL Analysis

```python
# Importing Liabraries

import pandas as pd
import sqlite3

print("Liabraries imported sucessfully!!")
```

```python
# loading dataset file

df = pd.read_csv("../data/cleaned/cleaned_netflix.csv")
print("File loaded!")
```

```python
# Creating netflix database

conn = sqlite3.connect("../sql/netflix.db")
print("DB Created!")
```

```python
# Connecitng with netflix database --> taking CSV data into netflix database

df.to_sql("netflix", conn, if_exists="replace",index=False)
```

```python
# Checking data available in netflix database

query = "SELECT * FROM netflix LIMIT 5"

pd.read_sql(query, conn)
```

Count Movies vs TV Shows

```python
query1 = """SELECT type, 
COUNT(*) AS total_content 
FROM netflix 
GROUP BY type"""

pd.read_sql(query1, conn)
```

TOP 10 Countries


```python
query2 = """ SELECT country,
COUNT(*) AS total_country
FROM netflix
GROUP BY country
ORDER BY total_country DESC
LIMIT 10"""

pd.read_sql(query2, conn)
```
